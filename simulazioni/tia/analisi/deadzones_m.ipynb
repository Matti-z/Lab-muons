```python
def count_hits(df):
    """
    Count events where Minerva and Partenope both == 1,
    and count events where all three detectors == 1.
    
    Parameters:
    df : pandas.DataFrame
        DataFrame with columns: HitBottom, HitTop, HitMiddle

    """
    count_both = ((df['HitBottom'] == 1) & (df['HitTop'] == 1)).sum()
    count_all_three = ((df['HitBottom'] == 1) & (df['HitTop'] == 1) & (df['HitMiddle'] == 1)).sum()
    return count_both,count_all_three

def filtering(a, b, c, d, peso_a = 0.54, peso_c = 0.54):
    filt1 = int(a * peso_a + b *(1-peso_a))
    filt2 = int(c * peso_c + d *(1-peso_c))
    return filt2 / filt1
```

```python
angles = [0, 15, 30, 45, 60]
variants = ['', 's', '1', '2']  # The suffixes: d0m, d0ms, d0m1, d0m2
polarities = ['m', 'p']

results = {}

for angle in angles:
    results[angle] = {}
    
    for variant in variants:
        # Get both m and p polarities for this angle/variant
        d_m = globals()[f'd{angle}m{variant}']
        t_m = globals()[f't{angle}m{variant}']
        d_p = globals()[f'd{angle}p{variant}']
        t_p = globals()[f't{angle}p{variant}']
        
        # Apply filtering (or store as-is, depending on your logic)
        results[angle][variant] = filtering(d_m, d_p, t_m, t_p)

print(results)

errors = {}

for angle in angles:
    errors[angle] = {}

    for variant in variants:
        d_m = globals()[f'd{angle}m{variant}']
        d_p = globals()[f'd{angle}p{variant}']
        t_m = globals()[f't{angle}m{variant}']
        t_p = globals()[f't{angle}p{variant}']

        filt1 = int(d_m * 0.54 + d_p * 0.46)
        filt2 = int(t_m * 0.54 + t_p * 0.46)

        errors[angle][variant] = frac([filt2, filt1])

print (errors)
```

```python
labels = {
    '': r'$\cos(\theta)$',
    's': r'$\cos^{2}(\theta)$',
    '1': r'$\cos^{1/3}(\theta)$',
    '2': r'$\cos^{2/3}(\theta)$'
}

y_M =  (df_M[' pgm_triple'] /df_M[' pg_doppie']).iloc[:5].values
err_M = np.array([
    frac(a, b)
    for a, b in zip(df_M[' pgm_triple'][:5],
                    df_M[' pg_doppie'][:5])
])

x = np.linspace(0, 60, 5)

plt.figure(figsize= (15,6))
for key in ['', 's', '1', '2']:
    y = [results[t][key] for t in x]
    err_y = [errors[t][key] for t in x]
    plt.errorbar(x, y, err_y, label=labels[key], fmt = '.')

plt.errorbar(x, y_M, yerr= err_M , label = 'Minerva', )
plt.ylabel(r'$ \epsilon $', fontsize = 20)
plt.xlabel('Misalignment [cm]', fontsize = 15)
plt.legend()
plt.grid(True, which='both', linestyle='--', linewidth=0.5, alpha=0.6,)

# plt.savefig('m_eff.pdf')
```

```python

# 1. Labels mapping matching your PDF legend
labels = {
    '': r'$\cos(\theta)$',
    's': r'$\cos^{2}(\theta)$',
    '1': r'$\cos^{1/3}(\theta)$',
    '2': r'$\cos^{2/3}(\theta)$'
}

# 2. Alignment distances (x-axis)
x = np.array([0, 15, 30, 45, 60])

# --- CALCULATING AND PLOTTING DISTANCES / SIGMA ---
plt.figure(figsize=(15, 6))

# Loop through each of the 4 theoretical models
for key in ['', 's', '1', '2']:
    # Extract model values and errors for the current key across all x points
    y_model = np.array([results[t][key] for t in x])
    sigma_model = np.array([errors[t][key] for t in x])
    
    # Convert y_M to a numpy array for vector operations
    y_experimental = np.array(y_M)
    
    # Calculate (Experimental - Model) / Sigma
    # Note: If you want absolute distance, change to: np.abs(y_experimental - y_model) / sigma_model
    normalized_distance = (y_experimental - y_model) / sigma_model
    
    # Plot the normalized distances as points connected by a dashed line
    plt.plot(x, normalized_distance, marker='o', linestyle='--', label=f"Distance to {labels[key]}")

# --- STYLING THE PLOT ---
plt.ylabel(r'Distance / $\sigma$', fontsize=18)
plt.xlabel('Misalignment [cm]', fontsize=16)
plt.xticks(fontsize=12)
plt.yticks(fontsize=12)

# Add a horizontal line at 0 to clearly show where the experimental data perfectly matches a model
plt.axhline(0, color='black', linestyle=':', alpha=0.5)

plt.grid(True, which='both', linestyle='--', linewidth=0.5, alpha=0.6)
plt.xlim(-3, 63)
plt.legend(loc='upper right', fontsize=12)

# plt.show()
# plt.savefig('m_eff_dist.pdf')
```
